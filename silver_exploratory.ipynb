# Explore & Transform Data Untuk Lakeflow Declarative Pipeline

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

---

## FACT TABLE

## bookings

```python
df = (
    spark.read
    .format("delta")
    .load("/Volumes/flyproject/bronze/bronze_volume/bookings/data/")
)

display(df)
```

```python
df = (
    df.withColumn("booking_date", to_date(col("booking_date")))
    .withColumn("amount", col("amount").cast(DoubleType()))
    .withColumn("processed_timestamp", current_timestamp())
    .drop("_rescued_data")
)

display(df.limit(5))
```

---

## DIMENSION TABLE

### flights

```python
df = (
    spark.read
    .format("delta")
    .load("/Volumes/flyproject/bronze/bronze_volume/flights/data/")
)

display(df)
```

```python
df = (
    df.withColumn("flight_date", to_date(col("flight_date")))
    .withColumn("processed_timestamp", current_timestamp())
    .drop("_rescued_data")
    .dropDuplicates()
)

display(df.limit(5))
```

### airports

```python
df = (
    spark.read
    .format("delta")
    .load("/Volumes/flyproject/bronze/bronze_volume/airports/data/")
)

display(df.limit(5))
```

```python
df = (
    df.withColumn("processed_timestamp", current_timestamp())
    .withColumn("airport_name", trim(col("airport_name")))
    .withColumn("city", trim(col("city")))
    .withColumn("country", trim(col("country")))
    .drop("_rescued_data")
    .dropDuplicates()
)

display(df.limit(5))
```

### passengers

```python
df = (
    spark.read
    .format("delta")
    .load("/Volumes/flyproject/bronze/bronze_volume/passengers/data/")
)

display(df.limit(5))
```

```python
df = (
    df.withColumn("processed_timestamp", current_timestamp())
    .withColumn("name", trim(col("name")))
    .withColumn("gender", trim(col("gender")))
    .withColumn("nationality", trim(col("nationality")))
    .drop("_rescued_data")
    .dropDuplicates()
)

display(df.limit(5))
```
